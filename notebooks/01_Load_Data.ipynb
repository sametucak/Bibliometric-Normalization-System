```python
def clean_text(text):
    """
    Metin temizleme fonksiyonu
    """

    if pd.isna(text):
        return ""

    text = str(text)

    text = text.strip()

    return text
```

```python
clean_text("   University of Parma   ")
```

```python
clean_text(None)
```

```python
for col in df.columns:
    if df[col].dtype == "object":
        df[col] = df[col].apply(clean_text)
```

```python
import re
import pandas as pd

def normalize_author(name):
    """
    Yazar ismini standartlaştırır.
    """

    if pd.isna(name):
        return ""

    name = str(name)

    # Baştaki ve sondaki boşlukları sil
    name = name.strip()

    # Birden fazla boşluğu teke indir
    name = re.sub(r"\s+", " ", name)

    # Büyük harfleri normale çevir
    name = name.title()

    return name
```

```python
normalize_author("  LUGLI,   GABRIELE ANDREA ")
```

```python
normalize_author("chen   yang")
```

```python
df["normalized_author"] = df["author_full_name"].apply(normalize_author)
```
